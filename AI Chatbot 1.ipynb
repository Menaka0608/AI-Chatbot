```python
print("Total questions:", len(data))

print("\nFirst 5 questions:")
for question, answer in data[:5]:
    print("Q:", question)
    print("A:", answer)
    print()
```

```python
questions = [item[0] for item in data]
answers = [item[1] for item in data]

print("Number of questions:", len(questions))
print("Number of answers:", len(answers))

print("\nSample question:", questions[0])
print("Sample answer:", answers[0])
```

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(questions)

sequences = tokenizer.texts_to_sequences(questions)

print("Vocabulary size:", len(tokenizer.word_index))
print("First question:", questions[0])
print("Converted sequence:", sequences[0])
```

```python
max_len = max(len(sequence) for sequence in sequences)

X = pad_sequences(
    sequences,
    maxlen=max_len,
    padding="post"
)

print("Maximum sentence length:", max_len)
print("Input shape:", X.shape)
print("\nFirst padded sequence:")
print(X[0])
```

```python
encoder = LabelEncoder()

y = encoder.fit_transform(answers)

print("Number of answer classes:", len(encoder.classes_))
print("First answer:", answers[0])
print("Encoded first answer:", y[0])
```

```python
vocab_size = len(tokenizer.word_index) + 1
num_classes = len(encoder.classes_)

model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=64),
    LSTM(128),
    Dense(num_classes, activation="softmax")
])

print("LSTM model created successfully!")
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    X,
    y,
    epochs=100,
    batch_size=8,
    verbose=1
)

print("Model training completed successfully!")
```

```python
def chatbot():
    print("====================================")
    print("        AI CHATBOT IS READY")
    print("====================================")
    print("Type 'exit' to stop the chatbot.")
    print()

    while True:
        user_input = input("You : ")

        if user_input.lower() == "exit":
            print("Bot : Goodbye! Have a nice day.")
            break

        sequence = tokenizer.texts_to_sequences([user_input])

        if not sequence[0]:
            print("Bot : Sorry, I don't understand that question yet.")
            continue

        sequence = pad_sequences(
            sequence,
            maxlen=max_len,
            padding="post"
        )

        prediction = model.predict(sequence, verbose=0)

        confidence = np.max(prediction)
        index = np.argmax(prediction)

        if confidence < 0.50:
            print("Bot : Sorry, I don't know the answer to that.")
        else:
            response = encoder.inverse_transform([index])[0]
            print("Bot :", response)
```
